```python
import numpy as np
```

```python
from sympy import Symbol, cos as c, sin as s, pi
```

```python
def transformation_matrix(alpha, a, d, theta):
    T = np.array([[c(theta),              -s(theta),             0,                             a],
                     [s(theta)*c(alpha),     c(theta)*c(alpha),    -s(alpha),             -s(alpha)*d],
                     [s(theta)*s(alpha),  c(theta)*s(alpha),        c(alpha),              c(alpha)*d],
                     [0,                         0,                  0,                             1]
                     ])
    return T
```

```python
theta1 = Symbol('theta1')
theta2 = Symbol('theta2')
theta3 = Symbol('theta3')
theta4 = Symbol('theta4')
theta5 = Symbol('theta5')
theta6 = Symbol('theta6')
a2 = Symbol('a2')
d3 = Symbol('d3')
a3 = Symbol('a3')
d4 = Symbol('d4')
```

```python
dh = np.array([[0, 0, 0, theta1],
              [-pi/2, 0, 0, theta2],
               [0, a2, d3, theta3],
               [-pi/2, a3, d4, theta4],
               [pi/2, 0, 0, theta5],
               [-pi/2, 0, 0, theta6]])
print(dh)
```

```python
T1_0 = transformation_matrix(dh[0][0], dh[0][1], dh[0][2], dh[0][3])

print(T1_0)
```

```python
T2_1 = transformation_matrix(dh[1][0], dh[1][1], dh[1][2], dh[1][3])

print(T2_1)
```

```python
T3_2 = transformation_matrix(dh[2][0], dh[2][1], dh[2][2], dh[2][3])

print(T3_2)
```

```python
T4_3 = transformation_matrix(dh[3][0], dh[3][1], dh[3][2], dh[3][3])

print(T4_3)
```

```python
T5_4 = transformation_matrix(dh[4][0], dh[4][1], dh[4][2], dh[4][3])

print(T5_4)
```

```python
T6_5 = transformation_matrix(dh[5][0], dh[5][1], dh[5][2], dh[5][3])

print(T6_5)
```

```python
T6_4 = np.matmul(T5_4, T6_5)

print(T6_4)
```

```python
T6_3 = np.matmul(T4_3, T6_4)

print(T6_3)
```

```python
T3_1 = np.matmul(T2_1, T3_2)

print(T3_1)
```

```python
T6_1 = np.matmul(T3_1, T6_3)

print(T6_1)
```

```python
T6_0 = np.matmul(T1_0, T6_1)

print(T6_0)
```
