```python
import numpy as np
```

```python
from sympy import Symbol, cos as c, sin as s, pi
```

```python
def transformation_matrix(alpha, a, d, theta):
    T = np.array([[c(theta),              -s(theta),             0,                             a],
                     [s(theta)*c(alpha),     c(theta)*c(alpha),    -s(alpha),             -s(alpha)*d],
                     [s(theta)*s(alpha),  c(theta)*s(alpha),        c(alpha),              c(alpha)*d],
                     [0,                         0,                  0,                             1]
                     ])
    return T
```

```python
theta1 = Symbol('theta1')
theta3 = Symbol('theta3')
d2 = Symbol('d2')
L2 = Symbol('L2')
```

```python
dh = np.array([[0, 0, 0, theta1],
              [pi/2, 0, d2, 0],
               [0, 0, L2, theta3]])
print(dh)
```

```python
# {1} relative to {0}
T01 = transformation_matrix(dh[0][0], dh[0][1], dh[0][2], dh[0][3])

print(T01)
```

```python
# {2} relative to {1}
T12 = transformation_matrix(dh[1][0], dh[1][1], dh[1][2], dh[1][3])

print(T12)
```

```python
# {3} relative to {2}
T23 = transformation_matrix(dh[2][0], dh[2][1], dh[2][2], dh[2][3])

print(T23)
```

```python
# {31} relative to {1}
T13 = np.matmul(T12, T23)

print(T13)
```

```python
# {3} relative to {0}
T03 = np.matmul(T01, T13)

print(T03)
```
